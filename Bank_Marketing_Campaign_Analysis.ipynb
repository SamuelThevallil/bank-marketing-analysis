**Subscription Rates Across DIfferent Groups**

```python
# 1 = subscribed, 0 = not subscribed

banking['subscribed'] = (banking['y'] == 'yes').astype(int)

overall_rate = banking['subscribed'].mean() * 100
print(f"Overall subscription rate: {overall_rate:.2f}%")

# Age is continuous, so bucket it into groups first

banking['age_group'] = pd.cut(
    banking['age'],
    bins=[0, 24, 34, 44, 54, 64, 120],
    labels=['<=24', '25-34', '35-44', '45-54', '55-64', '65+']
)
```

```python
def subscription_rate(col, sort=True):
    """Customers, subscribers and subscription rate (%) for each group in `col`."""
    out = (banking.groupby(col, observed=True)['subscribed']
                  .agg(customers='count', subscribers='sum', rate='mean'))
    out['rate'] = (out['rate'] * 100).round(2)
    return out.sort_values('rate', ascending=False) if sort else out


def plot_rate(col, sort=True, rotate=0):
    data = subscription_rate(col, sort)

    plt.figure(figsize=(10, 5))
    ax = sns.barplot(x=data.index.astype(str), y=data['rate'], color='steelblue')
    ax.bar_label(ax.containers[0], fmt='%.1f%%', padding=2)

    # dashed line = overall average, so you can see which groups beat it
    plt.axhline(overall_rate, color='red', linestyle='--',
                label=f'Overall: {overall_rate:.1f}%')

    plt.title(f'Subscription Rate by {col.replace("_", " ").title()}')
    plt.xlabel(col.replace("_", " ").title())
    plt.ylabel('Subscription Rate (%)')
    plt.xticks(rotation=rotate)
    plt.legend()
    plt.tight_layout()
    plt.show()
```

```python
# Age keeps its natural order (sort=False); the others are sorted by rate
print(subscription_rate('age_group', sort=False))
plot_rate('age_group', sort=False)

for col, rot in [('job', 45), ('education', 30), ('marital', 0),
                 ('housing', 0), ('loan', 0), ('default', 0)]:
    print(f"\n--- {col} ---")
    print(subscription_rate(col))
    plot_rate(col, rotate=rot)
```

**Analyze contact methods, campaign months, call duration, and number of
customer contacts.**

```python
from scipy.stats import chi2_contingency

# Put months and weekdays in calendar order so charts don't sort alphabetically

month_order = ['mar','apr','may','jun','jul','aug','sep','oct','nov','dec']
day_order   = ['mon','tue','wed','thu','fri']
banking['month']       = pd.Categorical(banking['month'], categories=month_order, ordered=True)
banking['day_of_week'] = pd.Categorical(banking['day_of_week'], categories=day_order, ordered=True)

# Call duration is in seconds -> group into minutes

banking['duration_group'] = pd.cut(
    banking['duration'], bins=[-1, 60, 180, 300, 600, np.inf],
    labels=['0-1 min', '1-3 min', '3-5 min', '5-10 min', '10+ min'])

# Number of contacts in this campaign -> group (a few customers were called 40+ times)

banking['campaign_group'] = pd.cut(
    banking['campaign'], bins=[0, 1, 2, 3, 5, 10, np.inf],
    labels=['1', '2', '3', '4-5', '6-10', '11+'])

# pdays = 999 means "never contacted before", so flag it instead of treating it as a number

banking['previously_contacted'] = np.where(banking['pdays'] == 999, 'No', 'Yes')
banking['previous_group'] = pd.cut(
    banking['previous'], bins=[-1, 0, 1, 2, np.inf], labels=['0', '1', '2', '3+'])
```

```python
# ---------- Contact method ----------

print(subscription_rate('contact'))
plot_rate('contact')

# ---------- Campaign month: volume (bars) vs rate (line) ----------

month_tbl = subscription_rate('month', sort=False)
print(month_tbl)

fig, ax1 = plt.subplots(figsize=(10, 5))
ax1.bar(month_tbl.index.astype(str), month_tbl['customers'], color='lightgray')
ax1.set_xlabel('Month')
ax1.set_ylabel('Customers contacted')
ax2 = ax1.twinx()
ax2.plot(month_tbl.index.astype(str), month_tbl['rate'], color='red', marker='o')
ax2.set_ylabel('Subscription rate (%)')
plt.title('Campaign Volume vs Subscription Rate by Month')
plt.show()

# ---------- Day of week ----------

print(subscription_rate('day_of_week', sort=False))
plot_rate('day_of_week', sort=False)

# ---------- Call duration ----------

print(banking.groupby('y')['duration'].describe().round(1))

sns.histplot(data=banking, x='duration', hue='y', bins=60,
             stat='density', common_norm=False, element='step')
plt.xlim(0, 1500)
plt.title('Call Duration (seconds) by Subscription Outcome')
plt.show()

print(subscription_rate('duration_group', sort=False))
plot_rate('duration_group', sort=False)

# ---------- Number of customer contacts ----------

sns.countplot(data=banking[banking['campaign'] <= 15], x='campaign', color='steelblue')
plt.title('Number of Contacts in This Campaign (up to 15)')
plt.xlabel('Contacts')
plt.ylabel('Customers')
plt.show()

print(subscription_rate('campaign_group', sort=False))
plot_rate('campaign_group', sort=False)
```

**Compare successful and unsuccessful campaign outcomes.**



```python
# Overall split of outcomes

print(banking['y'].value_counts())
print(banking['y'].value_counts(normalize=True).mul(100).round(2))
sns.countplot(data=banking, x='y')
plt.title('Subscribed (yes) vs Not Subscribed (no)')
plt.show()

# Numeric comparison: yes vs no

print(banking.groupby('y')[['age', 'duration', 'campaign', 'previous']]
             .agg(['mean', 'median']).round(2))

# pdays only makes sense for customers contacted before (exclude 999)

print(banking[banking['pdays'] != 999].groupby('y')['pdays']
             .agg(['count', 'mean', 'median']).round(1))

# Box plots (outliers hidden so the boxes are readable)

fig, axes = plt.subplots(1, 3, figsize=(15, 5))
for ax, col in zip(axes, ['duration', 'campaign', 'age']):
    sns.boxplot(data=banking, x='y', y=col, showfliers=False, ax=ax)
    ax.set_title(f'{col.title()} by Outcome')
plt.tight_layout()
plt.show()

# Categorical comparison: how do yes and no customers differ in contact method?

print((pd.crosstab(banking['y'], banking['contact'], normalize='index') * 100).round(2))
```

**Examine the relationship between previous campaign results and current
subscriptions.**

```python
# Subscription rate by previous outcome

print(subscription_rate('poutcome'))
plot_rate('poutcome')

# Counts of yes/no within each previous outcome

sns.countplot(data=banking, x='poutcome', hue='y')
plt.title('Current Subscription by Previous Campaign Outcome')
plt.show()

# Contacted before or not

print(subscription_rate('previously_contacted'))
plot_rate('previously_contacted')

# Number of earlier contacts

print(subscription_rate('previous_group', sort=False))
plot_rate('previous_group', sort=False)

# Is the relationship statistically significant? (chi-square test of independence)

chi2, p, dof, _ = chi2_contingency(pd.crosstab(banking['poutcome'], banking['y']))
print(f"Chi-square = {chi2:.1f}, p-value = {p:.4g}")
```

**Identify customer segments and campaign patterns associated with higher
subscription rates.**

```python
# Best / worst group for every feature, in one table # (groups with fewer than 100 customers are ignored so tiny groups don't distort it)

for col in ['age_group', 'job', 'education', 'marital', 'contact', 'month',
            'day_of_week', 'poutcome', 'duration_group', 'campaign_group']:
    t = subscription_rate(col)
    t = t[t['customers'] >= 100]
    print(f"{col:15s} highest: {str(t.index[0]):20s} {t['rate'].iloc[0]:6.2f}%   "
          f"lowest: {str(t.index[-1]):20s} {t['rate'].iloc[-1]:6.2f}%")


# Top combined segments

def top_segments(cols, min_size=100, n=10):
    seg = (banking.groupby(cols, observed=True)['subscribed']
                  .agg(customers='count', subscribers='sum', rate='mean'))
    seg['rate'] = (seg['rate'] * 100).round(2)
    seg = seg[seg['customers'] >= min_size]
    return seg.sort_values('rate', ascending=False).head(n)

print("\nTop age group x job segments");      print(top_segments(['age_group', 'job']))
print("\nTop job x contact segments");        print(top_segments(['job', 'contact']))
print("\nTop month x contact segments");      print(top_segments(['month', 'contact']))
print("\nTop poutcome x contact segments");   print(top_segments(['poutcome', 'contact'], min_size=50))


# Heatmaps of subscription rate for pairs of features

def rate_heatmap(row, col, min_size=50):
    rate = banking.pivot_table(index=row, columns=col, values='subscribed',
                               aggfunc='mean', observed=True) * 100
    size = banking.pivot_table(index=row, columns=col, values='subscribed',
                               aggfunc='count', observed=True)
    rate = rate.where(size >= min_size)          # hide cells with very few customers
    plt.figure(figsize=(11, 6))
    sns.heatmap(rate, annot=True, fmt='.1f', cmap='YlGnBu', linewidths=0.5,
                cbar_kws={'label': 'Subscription rate (%)'})
    plt.title(f'Subscription Rate (%): {row} x {col}')
    plt.show()

rate_heatmap('month', 'contact')
rate_heatmap('job', 'age_group')
rate_heatmap('poutcome', 'contact')
```
